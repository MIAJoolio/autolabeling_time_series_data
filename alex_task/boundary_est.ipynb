```python
# !pip install opencv-python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
bgr_image1 = cv2.imread('data/three_color_image.png')
rgb_image1 = cv2.cvtColor(bgr_image1, cv2.COLOR_BGR2RGB)

plt.imshow(rgb_image1)
```

```python
bgr_image2 = cv2.imread('data/rgb_map_original.png')
rgb_image2 = cv2.cvtColor(bgr_image2, cv2.COLOR_BGR2RGB)
gray_image2 = cv2.cvtColor(bgr_image2, cv2.COLOR_RGB2GRAY)

plt.figure(figsize=(24,24))
plt.imshow(rgb_image2)
```

```python
contours, hierarchy = cv2.findContours(gray_image2, mode=cv2.RETR_TREE, method=cv2.CHAIN_APPROX_SIMPLE)# [0][2]

print("Total Number of Contours found =", len(contours))
print("contours are: \n",contours)
print("hierarchy is: \n",hierarchy)

min_area = 100
filtered_contours = [cnt for cnt in contours if cv2.contourArea(cnt) > min_area]

rgb_image2_copy = rgb_image2.copy()

for cnt in filtered_contours:
    # approx = cv2.approxPolyDP(cnt, 0.04 * cv2.arcLength(cnt, True), True)
    # if len(approx) == 4:  # Ищем четырехугольники
    #     cv2.drawContours(rgb_image2_copy, [cnt], -1, (255, 0, 0), 2)
    cv2.drawContours(rgb_image2_copy, [cnt], -1, (255, 0, 0), 2)

plt.figure(figsize=(24,24))
plt.imshow(rgb_image2_copy)
```
